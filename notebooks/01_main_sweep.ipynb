the must-run sensitivity sweep (18 more). Each solve carries all three
weather scenarios in one problem. Standard CPU runtime, no GPU; about one
hour. Checkpoints are per (basis, blend) under `results/`, so a disconnect
only costs the unfinished case.

Run the cells in order. Rebuild the tables and figures afterwards with the
`src/build_*.py` scripts (see the README).
